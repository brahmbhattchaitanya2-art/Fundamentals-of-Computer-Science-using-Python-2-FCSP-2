```python
import pandas as pd
df=pd.read_csv("Ice_cream selling data.csv")
df

x=df[["Temperature (°C)"]]
y=df[["Ice Cream Sales (units)"]]
```

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error,r2_score
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
mse_list=[]
mae_list=[]
r2_list=[]

random_states =range(1,43)
for states in random_states:
    x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=states)
    poly =PolynomialFeatures(degree=2)
    x_poly_train=poly.fit_transform(x_train)
    x_poly_test=poly.transform(x_test)

    model=LinearRegression()

    model.fit(x_poly_train,y_train)


    y_pred=model.predict(x_poly_test)


    r2=r2_score(y_test,y_pred)
    mae=mean_absolute_error(y_test,y_pred)
    mse=mean_squared_error(y_test,y_pred)



    mse_list.append(mse)
    mae_list.append(mae)
    r2_list.append(r2)


    print(f"random_states : {states}")
    print(f"mean squared error : {mse_list}")
    print(f"mae{mae_list}")
    print(f"r2{r2_list}")

    #print("\t")
    # print("\n")

    


    
    
```
