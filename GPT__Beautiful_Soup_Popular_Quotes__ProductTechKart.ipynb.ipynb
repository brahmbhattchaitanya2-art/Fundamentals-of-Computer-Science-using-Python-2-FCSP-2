From the given Popular_Quotes.html, load the file ensuring the utf-8 encoding and do the following task using BeautifulSoup library in python.

1.	Extract the text from the <h1> tag and display it.
2.	Retrieve the text of the <title> tag, then display it in reverse order.
3.	What is the total number of <script> tags in the HTML?
4.	Extract all tags listed under the first quote by Oscar Wilde and print the tag with the most characters.
5.	Print the first 5 quotes along with their authors in the below given format:
Example:
Quote: “Be yourself; everyone else is already taken.” 
Author: Oscar Wilde
6.	Display the quotes that contain the word "accomplishment" in its text.
7.	Extract the number of likes for each quote and display them in ascending order.
8.	Find and print the names of authors who have more than one quote on the page.
9.	Create a DataFrame containing the quote, author name, and number of likes.

```python
import bs4
import pandas as pd
fh=open('Popular_Quotes.html',encoding='utf8')
soup= bs4.BeautifulSoup(fh,'html.parser')
```

```python
#1.	Extract the text from the <h1> tag and display it.
t=soup.find("h1").text.strip()
t
```

```python
#2.	Retrieve the text of the <title> tag, then display it in reverse order.
t=soup.find("title").text.strip()[::-1]
t
```

```python
#3.	What is the total number of <script> tags in the HTML?
tag3= soup.find_all('script')
print(len(tag3))
```

```python
#4.	Extract all tags listed under the first quote by Oscar Wilde and print the tag with the most characters.
tag4=soup.find('div', {'class':'greyText smallText left'}).text.strip().split()
maximum=0
for word in tag4:
    l=len(word)
    if l>maximum:
        maximum=l
        res=word
print(res)
```

```python
# 5.	Print the first 5 quotes along with their authors in the below given format:
# Example:
# Quote: “Be yourself; everyone else is already taken.” 
# Author: Oscar Wilde

qt=[]
au=[]

for tag in soup.find_all('div',{'class':'quoteText'}):
    d=tag.text.strip()
    st=d.find("“")
    end=d.find("”")
    
    qt.append(d[st+1:end])
# print(qt)
print("\n\n")

for author in soup.find_all('span',{'class':'authorOrTitle'}):
    a=author.text.strip()
    au.append(a)
# print(au)

for i in range(len(qt)):
    print("Quote: ", qt[i])
    print("Author: ",au[i])
    print("-------------")
```

```python
# 6.	Display the quotes that contain the word "accomplishment" in its text.
for i in qt:
    if 'accomplishment' in i:
        print(i)
```

```python
# 7.	Extract the number of likes for each quote and display them in ascending order.
likes=[]
for li in soup.find_all('a',{'class':'smallText'}):
    data=li.text.strip().split()[0]
    likes.append(data)
likes.sort()
print(likes)
```

```python
# # 8.	Find and print the names of authors who have more than one quote on the page.
# l=[]
# for name in au:
#     if au.count(name) and name not in l:
#         l.append(name)
# print(l)

l = []
for name in au:
    # Safely check if they appear more than once AND aren't already tracked
    if au.count(name) > 1 and name not in l:
        l.append(name)
print(l)
```

```python
# 9.	Create a DataFrame containing the quote, author name, and number of likes.

import pandas as pd

df=pd.DataFrame({"Quote":qt,
                "Author_name":au,
                "No_like":likes})
df
```

```python
from bs4 import BeautifulSoup
import pandas as pd

#Simulating the uploaded HTML structure

html_content = open("index (1).html",encoding="utf-8")
soup = BeautifulSoup(html_content, 'html.parser')
```

```python
# Level 1: Basic Extraction
# Q1. Extract the title of the webpage.

# Expected Output

# TechKart - Static Product Website

print(soup.title.text)
```

```python
# Q2. Extract the main heading of the website.
# TechKart


print(soup.header.h1.text)
```

```python
# Q3. Extract all navigation menu names.
# Home
# Products
# About
# Contact
# Amazon


navbar_links = soup.find('nav').find_all('a')
for link in navbar_links:
    print(link.text)
```

```python
# Q4. Extract all hyperlinks (href) present in the navbar.

# Output:

# #home
# #products
# #about
# #contact
# https://www.amazon.in




navbar_links = soup.find('nav').find_all('a')
for link in navbar_links:
    print(link['href'])
```

```python
# Level 2: Product Information
# Q5. Extract names of all products.

# Output:

# Wireless Mouse
# Mechanical Keyboard
# Bluetooth Speaker
# Smart Watch



products = soup.find_all('div', class_='product-card')
for card in products:
    print(card.find('h3', class_='product-name').text)
```

```python
# Q6. Extract prices of all products.

# Output:

# ₹799
# ₹2499
# ₹1599
# ₹3499


products = soup.find_all('div', class_='product-card')
for card in products:
    print(card.find('p', class_='price').text)
```

```python
# Q7. Extract ratings of all products.

# Output:

# 4.3 ★
# 4.7 ★
# 4.1 ★
# 4.5 ★



products = soup.find_all('div', class_='product-card')
for card in products:
    # Cleaning up the "Rating: " prefix using .replace()
    rating_text = card.find('p', class_='rating').text.replace("Rating: ", "")
    print(rating_text)
```

```python
# Q8. Extract seller names.

# Output:

# TechWorld
# KeyMaster
# SoundHub
# GadgetZone



products = soup.find_all('div', class_='product-card')
for card in products:
    seller_text = card.find('p', class_='seller').text.replace("Seller: ", "")
    print(seller_text)
```

```python
# Q9. Extract stock status of all products.

# Output:

# In Stock
# In Stock
# Out of Stock
# In Stock


products = soup.find_all('div', class_='product-card')
for card in products:
    print(card.find('p', class_='stock').text)
```

```python
# 2nd option to search 

# Safe approach: Only look inside the product cards
products = soup.find_all('div', class_='product-card')

for card in products:
    # This guarantees you only get the "View Deal" link for that specific product
    deal_link = card.find('a', target="_blank")
    print(deal_link['href'])
```

```python
# Level 4: Filtering Questions
# Q13. Display only products that are "In Stock".

# Expected:

# Wireless Mouse
# Mechanical Keyboard
# Smart Watch




products = soup.find_all('div', class_='product-card')
for card in products:
    stock_status = card.find('p', class_='stock')
    if 'available' in stock_status['class']:
        print(card.find('h3').text)
```

```python
# Q14. Display only products whose category is Accessories.

# Expected:

# Wireless Mouse
# Mechanical Keyboard



products = soup.find_all('div', class_='product-card')
for card in products:
    category = card.find('p', class_='category').text
    if "Accessories" in category:
        print(card.find('h3').text)
```

```python
# Q15. Find the most expensive product.

# Expected:

# Smart Watch
# ₹3499




products = soup.find_all('div', class_='product-card')
highest_price = 0
expensive_product = ""

for card in products:
    price_str = card.find('p', class_='price').text[1:]
    price_num = int(price_str)
    
    if price_num > highest_price:
        highest_price = price_num
        expensive_product = card.find('h3').text

print(expensive_product)
print(f"₹{highest_price}")
```

```python
# Q16. Find the cheapest product.

# Expected:

# Wireless Mouse
# ₹799




products = soup.find_all('div', class_='product-card')
lowest_price = float('inf') # Set initial value to infinity
cheapest_product = ""

for card in products:
    price_str = card.find('p', class_='price').text
    price_num = int(price_str.replace('₹', ''))
    
    if price_num < lowest_price:
        lowest_price = price_num
        cheapest_product = card.find('h3').text

print(cheapest_product)
print(f"₹{lowest_price}")
```

```python
# Q17. Find product with highest rating.

# Expected:

# Mechanical Keyboard
# 4.7




products = soup.find_all('div', class_='product-card')
highest_rating = 0.0
top_rated_product = ""

for card in products:
    rating_str = card.find('p', class_='rating').text.replace("Rating: ", "").replace(" ★", "")
    rating_num = float(rating_str)
    
    if rating_num > highest_rating:
        highest_rating = rating_num
        top_rated_product = card.find('h3').text

print(top_rated_product)
print(highest_rating)
```

```python
# Level 5: DataFrame Creation



# Q.18
# Create a DataFrame containing



products = soup.find_all('div', class_='product-card')
product_list = []

for card in products:
    data = {
        "Product ID": card['data-id'],
        "Name": card.find('h3').text,
        "Category": card.find('p', class_='category').text.replace("Category: ", ""),
        "Price": card.find('p', class_='price').text,
        "Rating": card.find('p', class_='rating').text.replace("Rating: ", ""),
        "Stock Status": card.find('p', class_='stock').text,
        "Seller": card.find('p', class_='seller').text.replace("Seller: ", ""),
        "Image URL": card.find('img')['src'],
        "Deal Link": card.find('a')['href']
    }
    product_list.append(data)

# Create DataFrame
df = pd.DataFrame(product_list)

# Display DataFrame
df
```
