```python
date=[]
stat=[]
tt=[]

for tag in soup.find_all('span',{'class':'short-desc'}):
    d=tag.strong.text.strip()
    date.append(d)
print(date)
```

```python
for tag in soup.find_all('span',{'class':'short-truth'}):
    d=tag.text.strip()
    tt.append(d[1:-1])
print(tt)
```

```python
for tag in soup.find_all('span',{'class':'short-desc'}):
    d=tag.text.strip()
    st=d.find('“')
    end=d.find('”')
    stat.append(d[st+1:end])
print(stat)
```

```python
print('No of records:',len(date))
print('==============')
for i in range(len(date)):
    print('Date: ',date[i])
    print('Statement: ',stat[i])
    print('Truth: ',tt[i])
    print('--------------')
```

```python
import pandas as pd
df=pd.DataFrame({'Date':date,
                 'Statement':stat,
                 'Truth Text':tt})
df
```

```python
import bs4
import pandas as pd
fh=open('Today 52 Week Low BSE_NSE Stocks Companies List – Ticker.html',encoding='utf8')
soup=bs4.BeautifulSoup(fh,'html.parser')
```

```python
comp=[]
price=[]
daylow=[]

for tag in soup.find_all('td',{'class':'left'}):
    data=tag.text.strip()
    comp.append(data)

for tag in soup.find_all('td',{'class':'Number'}):
    data=tag.text.strip()
    price.append(float(data))

for tag in soup.find_all('span',{'class':'Number'}):
    data=tag.text.strip()
    daylow.append(float(data))
    
df=pd.DataFrame({'Company':comp,
                 'Price in Rs.':price,
                 'Day Low':daylow})
df
```

## 4. Forecast air pollution data

```python
import requests,json
api_key="9e087baa4cae558ff2c7416885d00a12"
request_url=f"""http://api.openweathermap.org/data/2.5/air_pollution/forecast?lat={lat}&lon={long}&appid={api_key}"""
response=requests.get(request_url)
coordinates=response.json()
print(coordinates)
print(json.dumps(coordinates,indent=9,ensure_ascii=False))
```

```python
import pandas as pd
D={'date_time':[],'co':[],'no':[],'no2':[],'o3':[],'so2':[],'pm2_5':[],'pm10':[],'nh3':[]}
for i in coordinates['list']:
    D['date_time'].append(i['dt'])
    D['co'].append(i['components']['co'])
    D['no'].append(i['components']['no'])
    D['no2'].append(i['components']['no2'])
    D['o3'].append(i['components']['o3'])
    D['so2'].append(i['components']['so2'])
    D['pm2_5'].append(i['components']['pm2_5'])
    D['pm10'].append(i['components']['pm10'])
    D['nh3'].append(i['components']['nh3'])
df=pd.DataFrame(D)
df
```

# CRUD : create=>read=>update=>delete

```python
import requests
new_post={'title':'Learning Requests',
          'body':'This is tutorial content',
          'userId':1}
response=requests.post('https://jsonplaceholder.typicode.com/posts',json=new_post)
print(response.status_code)
print(response.json())
#201=created successfully
```

```python
#send put request
import requests
new_post={'id':1,
          'title':'updated title',
          'body':'updated body',
          'userId':1}
response=requests.put('https://jsonplaceholder.typicode.com/posts/1',json=new_post)
print(response.status_code)
print(response.json())
#201=created successfully
```

```python
#patch data
patch_data={'title':'Patched title'}
response=requests.patch('https://jsonplaceholder.typicode.com/posts/1',json=patch_data)
print(response.status_code)
print(response.json())
#201=created successfully
```

```python
#delete data
response=requests.delete('https://jsonplaceholder.typicode.com/posts/1')
print(response.status_code)
```
