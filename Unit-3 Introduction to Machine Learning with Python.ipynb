# get_dummies(data, columns, prefix=None, drop_first=False)
The get_dummies function is used to convert categorical variables into dummy or indicator variables. A dummy or indicator variable can have a value of 0 or 1.

```python
import pandas as pd
s = pd.Series(['A','X','E','B','D'])
res = pd.get_dummies(s)
print(res)
```

```python
import pandas as pd
s = pd.Series(['A','X','E','B','D'])
res = pd.get_dummies(s, drop_first=True, prefix='AA' )
print(res)
```

```python
data = {
'Area': [500, 750, 1000, 1200, 1500],
'Price': ['Low', 'Low', 'MId', 'High', 'High']
}
df = pd.DataFrame(data)

res = pd.get_dummies(df, prefix='AA' )
print(res)
```

```python
data = {
'Area': [500, 750, 1000, 1200, 1500],
'Price': ['Low', 'Low', 'MId', 'High', 'High'],
'Category':['A','B','A','B','A']
}
df = pd.DataFrame(data)

res = pd.get_dummies(df, prefix='AA' ,columns=['Price'])
print(res)
```

# np.where(condition, value_if_true, value_if_false)
numpy.where() is used for conditional selection and replacement in NumPy arrays. It can be used to:
1. Find indices that satisfy a condition
2. Build a new array by choosing values from two options depending on a condition.

```python
import numpy as np
arr = np.array([10, 15, 20, 25, 30])
result = np.where(arr > 20)
print(result)
```

```python
data = {'Marks': [10, 15, 20, 25, 30]}
df= pd.DataFrame(data)
df['category'] = np.where(df['Marks'] > 20, 'High',np.where(df['Marks'] <20, 'Low', 'Medium'))
print(df)
```

### 190 You've been given a dataset with apartment area and price information. There's a noticeable non-linear relationship between area and price. To address this you intend to categorize them into 'High', 'Medium', and 'Low' groups. Prices above 3,000,000 are 'High', below 2,000,000 are 'Low', and between 2,000,000 and 3,000,000 are 'Medium'. Write a code to achieve this assuming that dataset has two columns named area and price.

```python
import pandas as pd
import numpy as np
a = { 'Area': [500, 750, 1000, 1200, 1500],
'Price': [1500000, 2500000, 3000000, 3500000, 1800000]
}
df= pd.DataFrame(a)
df['Category'] = np.where(df['Price'] > 3000000, 'High',np.where(df['Price'] <2000000, 'Low', 'Medium'))
print(df)
```
