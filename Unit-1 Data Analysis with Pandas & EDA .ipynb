```python
import numpy as np
import pandas as pd

data = {
    "dept": ["it", "hr", "it", "hr", "sales"],
    "age": [25, 30, 28, 32, 40],
    "salary": [5000, 4000, 6000, 4500, 7000]
}
df = pd.DataFrame(data)

# compute quartiles and IQR for the "age" column
Q1 = df["age"].quantile(0.25)
Q3 = df["age"].quantile(0.75)
IQR = Q3 - Q1

low = Q1 - 1.5 * IQR
upp = Q3 + 1.5 * IQR

# use boolean mask with bitwise operators and proper DataFrame indexing
outliers = df[(df["age"] < low) | (df["age"] > upp)]

print("Outliers:")
print(outliers)
```

```python
Remove_outliers=df[(df["age"] > low) & (df["age"] < upp)]
```

```python
import seaborn as sns
sns.boxplot(data=outliers,x="age")
```

```python
import seaborn as sns
sns.boxplot(data=Remove_outliers,x="age")
```
