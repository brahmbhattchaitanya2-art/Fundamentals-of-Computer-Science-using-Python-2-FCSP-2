```python
F=soup.find_all("div",{"quoteText"})
ss=[]
for h in F:
    stm=h.text.strip()
    
    st=stm.find("“")
    end=stm.find("”")
    ss.append(stm[st+1:end])

auth=[]
for au in soup.find_all("span",class_="authorOrTitle"):
    a=au.text.strip()
    auth.append(a)
for i in range(5):
    print("Quote:",ss[i])
    print("Author:",auth[i])
    print("-----------------")
```

```python
products = Soup.find_all('div', class_='product-card')

lowest_price = 999999
lowest_product = ""

for card in products:
    name = card.find('h3').text
    price_str = card.find('p', class_='price').text
    price = int(price_str[1:])   # remove ₹

    if price < lowest_price:
        lowest_price = price
        lowest_product = name

print(lowest_product)
print("₹" + str(lowest_price))
```

```python
# Q17. Find product with highest rating.

# Expected:

# Mechanical Keyboard
# 4.7




products = Soup.find_all('div', class_='product-card')
highest_rating = 0.0
top_rated_product = ""

for card in products:
    rating_str = card.find('p', class_='rating').text[7:11]
    rating_num = float(rating_str)
    
    if rating_num > highest_rating:
        highest_rating = rating_num
        top_rated_product = card.find('h3').text

print(top_rated_product)
print(highest_rating)
```
